```python
order_list = []
print("請輸入您要點的餐點，輸入'/'結束：")
while True:
    item = input().strip()
    if item == '/':
        break
    if item:
        order_list.append(item)
has_salad = False
while "沙拉" in order_list:
    order_list.remove("沙拉")
    has_salad = True
if has_salad:
    print("今日無沙拉，已自動移除。")
order_list.append("冰淇淋")
print("完整訂單清單：")
print(", ".join(order_list))
```
